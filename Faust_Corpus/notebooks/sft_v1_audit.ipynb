```python
from pathlib import Path
import json, collections, hashlib, subprocess, sys
base = Path.cwd()
while not (base / 'data/master_segments.jsonl').exists():
    if (base / 'Faust_Corpus/data/master_segments.jsonl').exists():
        base = base / 'Faust_Corpus'; break
    if base == base.parent:
        raise FileNotFoundError('请从Faust_Corpus仓库内启动笔记本。')
    base = base.parent
rows = [json.loads(s) for s in (base/'data/sft_v1/faust_sft_pairs.jsonl').read_text(encoding='utf-8').splitlines()]
len(rows), collections.Counter(r['split'] for r in rows)
```
