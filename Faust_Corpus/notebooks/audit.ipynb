# 《浮士德》语料独立校验
只读取包内原文与数据，不调用模型或访问网络。下列代码运行 scripts/audit.py，核对每个正文事件、切分覆盖、集合隔离与复核记录。

```python
from pathlib import Path
import json, subprocess, sys
base = Path.cwd()
if base.name == "notebooks":
    base = base.parent
assert (base / "scripts" / "audit.py").exists(), "请从文件包根目录或notebooks目录运行"
result = subprocess.run([sys.executable, str(base / "scripts" / "audit.py"), "--base", str(base)], text=True, capture_output=True, check=True)
print(result.stdout)
```

```python
report = json.loads((base / "verification" / "quality_report.json").read_text("utf-8"))
assert report["all_checks_passed"]
report["source_statistics"], report["corpus_statistics"]
```

```python
boundaries = [json.loads(line) for line in (base / "review" / "boundary_review.jsonl").read_text("utf-8").splitlines()]
assert len(boundaries) == report["corpus_statistics"]["segments"]
[(b["segment_id"], b["reason_zh"]) for b in boundaries[:8]]
```
